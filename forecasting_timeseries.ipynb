```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("exchange rate dataset.csv")
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y %H:%M:%S")

# Monday is 0 and Sunday is 6.
df = df.loc[df["Date"].dt.dayofweek < 5].reset_index(drop=True)
print("Date range:", df["Date"].min(), "to", df["Date"].max())
print("Observations:", len(df))
print("Missing values:\n", df.isna().sum())
print("Duplicate dates:", df["Date"].duplicated().sum())
print(df.describe())
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df.groupby('Date')['Close'].mean(), label='average', color='red')
plt.title('Exchange Rate CNY/IDR Time Series')
plt.legend();
```
